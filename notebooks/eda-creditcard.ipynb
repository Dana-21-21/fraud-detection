```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")

import warnings
warnings.filterwarnings("ignore")
```

```python
credit_df = pd.read_csv("../data/raw/creditcard.csv")

credit_df.head()
```

```python
credit_df.shape
```

```python
credit_df.info()
```

```python
credit_df.isnull().sum()
```

```python
credit_df.duplicated().sum()
```

```python
credit_df = credit_df.drop_duplicates()
```

```python
credit_df.describe()
```

```python
class_dist = pd.DataFrame({
    "Count": credit_df["Class"].value_counts(),
    "Percentage":
    credit_df["Class"].value_counts(normalize=True)*100
})

class_dist
```

```python
plt.figure(figsize=(6,4))

sns.countplot(
    data=credit_df,
    x="Class"
)

plt.title(
    "Credit Card Fraud Distribution"
)

plt.show()
```

```python
fraud_rate = (
    credit_df["Class"].mean()
) * 100

print(f"Fraud Rate: {fraud_rate:.4f}%")
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    credit_df["Amount"],
    bins=50,
    kde=True
)

plt.title(
    "Transaction Amount Distribution"
)

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=credit_df,
    x="Class",
    y="Amount"
)

plt.title(
    "Transaction Amount by Fraud Class"
)

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    credit_df["Time"],
    bins=50
)

plt.title(
    "Transaction Time Distribution"
)

plt.show()
```

```python
corr = credit_df.corr()

fraud_corr = (
    corr["Class"]
    .sort_values(ascending=False)
)

fraud_corr.head(15)
```

```python
plt.figure(figsize=(15,10))

sns.heatmap(
    credit_df.corr(),
    cmap="coolwarm",
    center=0
)

plt.title(
    "Feature Correlation Heatmap"
)

plt.show()
```
